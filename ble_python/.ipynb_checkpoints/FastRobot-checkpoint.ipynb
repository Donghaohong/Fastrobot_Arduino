```python
tof_time = []
tof_front = []
tof_right = []

imu_time = []
pitch = []
roll = []

mode = None   

def handler(uuid, byte_array):
    global mode

    s = ble.bytearray_to_string(byte_array).strip()

    if not s:
        return

    if s.startswith("TOF"):
        mode = "TOF"
        return

    if s.startswith("IMU"):
        mode = "IMU"
        return

    if s.startswith("DONE"):
        print("Log finished")
        return

    parts = s.split(",")

    try:
        if mode == "TOF":
            t = int(parts[0])
            f = int(parts[1])
            r = int(parts[2])
            tof_time.append(t)
            tof_front.append(f)
            tof_right.append(r)

        elif mode == "IMU":
            t = int(parts[0])
            p = float(parts[1])
            rr = float(parts[2])
            imu_time.append(t)
            pitch.append(p)
            roll.append(rr)

    except:
        pass  

ble.start_notify(ble.uuid['RX_STRING'], handler)

ble.send_command(CMD.DATA_IMU_TOF, "")
```

```python
print("TOF samples:", len(tof_time))
print("IMU samples:", len(imu_time))
```

```python
ble.stop_notify(ble.uuid['RX_STRING'])
```
